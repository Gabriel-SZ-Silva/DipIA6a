10. Sequência de Fibonacci

Imprima os primeiros 10 números da sequência de Fibonacci.

```
# Décimo Exercício

# Cria a função para o código todo
def fibonnaciExe10(n):
  # Aqui cria e define duas variáveis, que são os dois primeiros digitos da sequência
  a = 0
  b = 1
  # Aqui se cria a lista que vai conter os números da sequência de fibonnaci
  seqExe10 = []
  # Cria uma repetição que se repete n vezes, para fazer o cálculo necessário para a sequência de fibonnaci
  for i in range(n):
    # Salva o valor a que é o resultado da soma dos dois números anteriores na lista seqExe10
    seqExe10.append(a)
    # Faz o cálculo, fazendo o cálculo no lado direito com os valores atuais, e então atualiza a parte do lado esquerdo
    # E então se repete n vezes
    a, b = b, a+b
    # Fecha a repetição com return
  return seqExe10

# Imprime a lista com os números da sequência
print(fibonnaciExe10(10))
```

21. Busca Binária

Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```
# Vintésimo Primeiro Exercício

# Criamos as variáveis para o código, primeiro a lista de números para fazermos a busca
# Depois, o índice inicial (que será 0)
# E o índice final, que usamos a função len e subtraimos por 1, para pegar o último número da lista
# E por fim, o número alvo digitado pelo usuário
listaExe21 = [10, 20, 30, 40, 50, 60]
iniExe21 = 0
fimExe21 = len(listaExe21) - 1
alvoExe21 = int(input("Digite o número: "))

 # Usamos a função while para criar um laço de repetição
 # Enquanto o índice final for maior ou igual ao inicial
while fimExe21 >= iniExe21:
  # Para começar a busca binária, criamos o índice do meio, que será definido pela divisão da soma do índice incial com o final
  # E dividido por 2, mas também devemos usar // para que forcemos o resultado a ser um valor inteiro
    meioExe21 = (iniExe21+fimExe21) // 2
    ## Caso o alvo e o valor do índice do meio sejam o mesmo, o sistema vai finalizar e avisar que encontrou
    # Então o break fará o loop finalizar
    if alvoExe21 == listaExe21[meioExe21]:
        print("Número encontrado")
        break
        # Caso o valor do alvo seja maior que o valor do índice do meio, o novo índice inicial será o índice do meio +1
        # Para que a lista seja "reduzida" e repetir o loop
    elif alvoExe21 > listaExe21[meioExe21]:
        iniExe21 = meioExe21 + 1
        # Caso o valor do alvo seja menor que o valor do índice do meio, o novo índice final será o indice do meio -1
        # Para que a lista seja reduzida e repetir o loop
    elif alvoExe21 < listaExe21[meioExe21]:
        fimExe21 = meioExe21 -1
        # O sistema caso detecte que o alvo nem está na lista, vai avisar e encerrar
else:
    print("Número não está na lista")
```

22. Criptografia de Cifra de César

Crie uma função para criptografar uma mensagem usando a cifra de
César.

```
# Vintésimo Segundo Exercício

# Além do clássico input, coloquei um .lower() para impedir a complicação de lidar com palavras maiusculas
# A primeira variável vai ser para a frase a ser criptografada, e a segunda será a chave da criptografia
txtExe22 = input('Insira a frase ou palavra para criptografar: ').lower()
chaveExe22 = int(input('Insira a chave da criptografia: '))

# Definimos uma função para a criptografia, usando com os parametros da frase e achave
def crip_cesar(txtExe22, chaveExe22):
  # Essa variável serve para armazenar a frase criptografada
    resultExe22 = ''
    # Criamos um laço de repetição para fazer a criptografia de cada letra
    for letraExe22 in txtExe22:
      # Usamos uma função isaplha() para verificar se um caractere realmente é uma letra
        if letraExe22.isalpha():
          # Usando a função ord() - que pega um valor e retorna seu valor correspondente na tabela ASCII -
          # Nós podemos fazer com que a letras se movam de acordo com a chave
          # Encarei o problema de que a tabela ASCII é diferente do alfabeto, então para corrigir
          # Devemos reduzir a tabela usando a letra "a" no ASCII, usando-a para subtrair da letra atual
          # Depois, devemos dividir por 26 e o resto será a nova posição da letra
          # E então, temos a nova posição da letra e a colocamos como letra novamente
            posExe22 = ord(letraExe22) - ord('a')
            novaPosExe22 = (chaveExe22 + posExe22) % 26
            codigo_final = novaPosExe22 + ord('a')
            resultExe22 = resultExe22 + chr(codigo_final)
        # Caso não seja uma letra mesmo, o mantém
        else:
            resultExe22 = resultExe22 + letraExe22
            # Após mudar a posição de cada letra e retornar, retornamos o resultado todo
    return resultExe22

print(crip_cesar(txtExe22, chaveExe22))
```

24. Validação de CPF

Implemente uma função para validar um CPF brasileiro.

```

# Vintésimo Quarto Exercício

cpfExe24 = input('Insira seu cpf: ')
def valida_cpf(cpfExe24):
    # Substituimos pontos e traços por nada
    cpfExe24 = cpfExe24.replace('.', '').replace('-', '')
    # Se tiver menos ou mais que 11, dá erro
    if len(cpfExe24) != 11:
        return False
        # Se todos os números forem iguais, dá erro
        # Para fazer isso, multiplica o primeiro numero do cpf e multiplica por 11
        # Então, cria uma string repetindo aquele número 11 vezes e verifica
    elif cpfExe24 == cpfExe24[0] * 11:
        return False
    somaExe24 = 0
    # Os 2 últimos números são gerados a partir dos 9 primeiros, então o laço envolve apenas eles
    for i in range(9):
        # O 10o número é gerado pegando o primeiro número e multiplicando por 10
        # Então o segundo é multiplicado por 9, e assim em diante
        somaExe24 += int(cpfExe24[i]) * (10 - i)
        # O resultado disso é o primeiro dígito
    digito1Exe24 = ((somaExe24 * 10) % 11)
    if digito1Exe24 == 10:
        digito1Exe24 = 0
        # Se o dígito verificador calculado pelo programa for diferente
        # do que provido pelo usuário, dá como falso
    if digito1Exe24 != int(cpfExe24[9]):
        return False
        # Retorna o soma ao 0 pra ser reutilizado
    somaExe24 = 0
    for i in range(10):
        # Fazemos o mesmo que o dígito verificador 1, mas desta vez
        # o incluindo no cálculo
        somaExe24 += int(cpfExe24[i]) * (11 - i)
    digito2Exe24 = ((somaExe24 * 10) % 11)
    if digito2Exe24 == 10:
        digito2Exe24 = 0
    if digito2Exe24 != int(cpfExe24[10]):
        return False
    return True
if valida_cpf(cpfExe24):
    print("CPF Válido!")
else:
    print("CPF Inválido!")
```

31. Ordenação por Bolha

```
#  Trigésimo Primeiro Exercício

# Primeiro criamos a lista a ser ordenada
lisExe31 = [14, 2, 98, 64, 11, 32, 77, 25]

# Em seguida criamos o primeiro laço de repetição
# para que o código passe por todos os elementos da lista
for i in range(len(lisExe31)):
    # Como base, a variável que informa se o elemento trocou
    # começa como falso
    trocExe31 = False
    # Depois criamos nosso segundo laço de repetição
    # que compara os elementos vizinhos
    for u in range(0, len(lisExe31) - i -1):
        # Essa condicional verifica se o elemento atual é maior que o vizinhos
        if lisExe31[u] > lisExe31[u + 1]:
            # Se for, então eles trocam de lugar
            lisExe31[u], lisExe31[u + 1] = lisExe31[u + 1], lisExe31[u]
            # E sinaliza que a troca foi feita
            trocExe31 = True
            # Caso não haja mais trocas, então a lista está feita
    if not trocExe31:
        break
print(f'Nova Lista Ordenada: {lisExe31}')
```

32. Jogo da Velha

Crie um jogo da velha para dois jogadores.

```
#  Trigésimo Segundo Exercício
# Criamos os dois jogadores e seus símbolos
p1Exe32 = 'X'
p2Exe32 = 'O'
# Aqui criamos uma variável para contagem de turnos
turno = 1
# Essa variável cria uma matriz para o tabuleiro
mtrExe32 = [['_','_','_'], ['_','_','_'], ['_','_','_']]
# Esta função serve para imprimir o tabuleiro corretamente
def tablr():
    # Essa impressão imprime os eixos x e y
    print("\n  0   1   2")
    # Faz o quadro de jogo de velha
    for i, linha in enumerate(mtrExe32):
        print(f"{i} " + " | ".join(linha))
        if i < 2:
            print("  ---------")
# Essa função faz a verificação se o jogo acabou como
# vitória para algum jogador
def ver_vitoria():
    # Faz a verificação se as linhas e colunas foram preenchidas
    for i in range(3):
        if mtrExe32[i][0] == mtrExe32[i][1] == mtrExe32[i][2] != '_':
            return True
        if mtrExe32[0][i] == mtrExe32[1][i] == mtrExe32[2][i] != '_':
            return True
    # Faz a verificação se alguma diagonal foi concluida
    if mtrExe32[0][0] == mtrExe32 [1][1] == mtrExe32 [2][2] != '_':
        return True
    if mtrExe32[0][2] == mtrExe32 [1][1] == mtrExe32 [2][0] != '_':
        return True
        # Caso não passe por nenhuma condição de vitória
        # significa que ainda não acabou a partida como vitória
    return False

    # Fazemos uma função para definir se a partida acaba como empate
def ver_empate():
    # Caso todas as linhas estejam apenas com o traço, retorna falso
    for linha in mtrExe32:
        if "_" in linha:
            return False
    return True

# Agora sim fazemos o jogo começar
while True:
    tablr()
    if turno == 1:
        pAtual = p1Exe32
    else:
        pAtual = p2Exe32
    print(f'\nTurno do jogador {pAtual}')
    jog_linha = int(input("Escolha a linha (0, 1 ou 2): "))
    jog_coluna = int(input("Escolha a coluna (0, 1 ou 2): "))
    if mtrExe32[jog_linha][jog_coluna] == '_':
        mtrExe32[jog_linha][jog_coluna] = pAtual
        if ver_vitoria():
            tablr()
            print(f'\nFim de Jogo, o jogador {pAtual} venceu.')
            break
        if ver_empate():
            tablr()
            print('\nFim de jogo, deu velha.')
            break
        # Aqui fazemos uma condicional para alterar o turno (se era 1 vira 2, se era 2 vira 1)
        if turno == 1:
            turno = 2
        else:
            turno = 1
    else:
        print('\n Posição já escolhida, escolha outra.')
```

47. Contar Números em Lista

```
# Quadragésimo Sétimo Exercício

lisExe47 = [7,7,12,59,65,12,8,8,8,4]
# esse dicionário mantém o número e quantas vezes aparece como chave e valor respectivamente.
qntExe47 = {}
# Um laço de repetição que percorre todos os números na lista lisExe47
for i in lisExe47:
    # E faz uma verificação para cada número, se não estiver no dicionario
    # ele o adiciona como chave, e deixa com valor 1
    if i not in qntExe47:
        qntExe47[i] = 1
        # caso ja esteja, apenas adiciona +1 no valor
    else:
        qntExe47[i] += 1
# Esse laço faz com que o valor seja definido como "valor" e chave seja definida como "chave" como tuplas
# e então imprime para cada item do dicionario
for chave, valor in qntExe47.items():
    print(f'O número {chave} aparece {valor} vez(es) na lista')
```

48. Ordenação por Inserção

Implemente o algoritmo de ordenação por inserção (Insertion Sort).


```
# Quadragésimo Oitavo Exercício

lisExe48 = [8,1,7,5,2,6]
print(f'Lista Original: {lisExe48}')
# Criamos esse laço de repetição para percorrer todos os indices da lista
# começando a partir do segundo, pois supõe-se que o primeiro já esteja ordenado e é nosso ponto de partida
# tambem usamos len() para que faça isso corretamente até o fim
for i in range(1, len(lisExe48)):
    # guardamos o valor do índice atual na variavel keyExe48
    keyExe48 = lisExe48[i]
    # e para fazer a ordenação por inserção, criamos uma variavel j que vai ser o elemento da esquerda do indice atual
    j = i - 1
    # esse laço empurra os números maiores para a direita enquanto o j for maior ou igual a 0
    # e enquanto for maior que o valor do indice atual
    while (j >= 0) and (lisExe48[j] > keyExe48):
        # isso empurra o número maior para a direita
        lisExe48[j + 1] = lisExe48[j]
        # E então olhamos pro proximo numero a esquerda
        j -= 1
    lisExe48[j + 1] = keyExe48
print(f'\nLista ordenada por inserção: {lisExe48}')
```

49. Soma dos Dígitos

Calcule a soma dos dígitos de um número inteiro fornecido.

```
# Quadragésimo Nono Exercício

# Embora o usuário digite um número, o código vai o ler como uma string, por ser mais fácil
# que manipular o número como um número seria mais complexo e custaria mais memória
strExe49 = input('Insira o número: ')
# essa variável armazena a soma dos dígitos
sumExe49 = 0
# esse laço percorre cada caractere da "palavra", que neste caso é um "número"
for i in strExe49:
    # essa linha soma o dígito atual ao total da soma, usando int() pra colocar o caractere
    # (que o sistema entende como letra) como um número real
    sumExe49 += int(i)
print(f'\nA soma dos dígitos do número {strExe49} é: {sumExe49}')
```

50. Cálculo de Frequência de Letras

Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```
# Quinquagésimo Exercício

# Além de pedir para o usuário inserir a frase, uso a função .upper() para deixar todas as letras
# em caixa alta, também seria possível usar .lower() para deixar em caixa baixa, mas na hora de exibir
# acho melhor esteticamente, e tem melhor visualização, se for em caixa alta.
# é necessário transformar tudo em caixa alta ou baixa para evitar que o programa identifique letras em caixas
# diferentes como letras diferentes.
# Também se usa replace para não identificar espaços como letras
strExe50 = input('Insira a frase: ').upper().replace(' ','')
# esse dicionário mantém a letra e seu valor como chave e valor respectivamente.
qntExe50 = {}
# Um laço de repetição que percorre toda a string strExe50
for i in strExe50:
    # E faz uma verificação para cada caractere, se não estiver no dicionario
    # ele o adiciona como chave, e deixa com valor 1
    if i not in qntExe50:
        qntExe50[i] = 1
    else:
        # caso ja esteja, apenas adiciona +1 no valor
        qntExe50[i] += 1
# Esse laço faz com que o valor seja definido como "valor" e chave seja definida como "chave" como tuplas
# e então imprime para cada item do dicionario
for chave, valor in qntExe50.items():
    print(f'A letra {chave} aparece {valor} vez(es) na frase')
```
