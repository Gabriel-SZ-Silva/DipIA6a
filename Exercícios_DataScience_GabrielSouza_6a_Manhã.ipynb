```python
# Primeiro Exercício
import numpy as np
# importamos scipy para estes exercicios, porém como ele
# é gigantesco, importamos submódulos especificos dele, neste caso
# sendo o integrate
import scipy.integrate as spi

# criamos uma função para o cálculo básico, usando x como parametro
def f(x):
    # e fazemos a função retornar o resultado da fórmula
    return (x**2) * np.sin(x)
# e então usamos a função .quad definindo os limites inferiores e superiores
# e especificamos qual função iremos usar o quad para cálcular a integral definida de uma única variável
resultExe01SP = spi.quad(f, 0, np.pi)
print(resultExe01SP)
```

4. Use a função de otimização de SciPy para maximizar f(x)= −x^2+4x+1.

```python
# Quarto Exercício
# importamos o scipy para estes exercícios, mas como ele
# é gigantesco, importamos submódulos específicos dele, neste caso
# sendo o optimize
import scipy.optimize as spo

# embora o enunciado nos peça para maximizar a fórmula, para fazer isso
# no Python usando o SciPy, devemos usar a função que a minimiza. Para então maximizar,
# usamos um truque invertendo o sinal da fórmula, para que na hora que a função minimize
# encontrar o ponto mínimo, ele seja exatamente o máximo da original
def f04(x):
    return -( -x**2 + 4*x + 1 )

# agora ao chamar a função minimize para minimizar a nossa função, definimos um palpite
# inicial de 0. Devido à forma como a fórmula está invertida, essa minimização vai encontrar
# exatamente o valor que maximiza a nossa equação original
resultExe04 = spo.minimize(f04, 0)
print(resultExe04)
```

10. Calcule a integral dupla da função f(x, y) = xy + x
2 no intervalo [0, 2] para x e [0, 1] para y.


```python
# Décimo Exercício
# importamos o submódulo integrate do scipy para realizar cálculos de integração numérica,
# técnica fundamental em ciência de dados para calcular volumes, probabilidades acumuladas e áreas
import scipy.integrate as spi

# definimos a função matemática f(x, y) = xy + x².
# de acordo com as regras de ouro do dblquad, a variável interna de integração (y)
# deve ser obrigatoriamente colocada como o primeiro parâmetro, seguida da externa (x)
def fExe10SP(y, x):
    return x*y + x**2

# chamamos a função dblquad para calcular a integral dupla.
# passamos a função, os limites numéricos da variável externa x (de 0 a 2), e definimos
# os limites da variável interna y através de funções lambda simples (de 0 a 1)
resultExe10SP = spi.dblquad(fExe10SP, 0, 2, lambda x: 0, lambda x: 1)

# exibimos o resultado do cálculo. o retorno será uma tupla contendo o valor aproximado
# da integral (3.6666...) e a estimativa de erro absoluto do algoritmo
print(resultExe10SP)
```

16. Resolva a equação diferencial de segunda ordem d2y/dx2 +y0, com condições
iniciais y(0) = 1 e y′(0) = 0.

```python
# Décimo Sexto Exercício
# importamos o submódulo integrate do scipy para resolver equações diferenciais ordinais (EDOs),
# ferramenta indispensável para modelar sistemas dinâmicos e físicos em computação científica
import scipy.integrate as spin

# definimos o sistema de primeira ordem equivalente à nossa EDO de segunda ordem y'' + y = 0.
# o parâmetro 'x' representa a variável independente (tempo/espaço) e 'v' armazena o vetor de estado [y, y']
def sisExe16(x, v):
    y0, y1 = v        # y0 é a posição y, y1 é a velocidade y' (recebe as duas condições iniciais do vetor)
    dy0_dx = y1       # a derivada da posição é a velocidade (y')
    dy1_dx = -y0      # pela EDO original (y'' = -y), a derivada da velocidade é -y0
    return [dy0_dx, dy1_dx]

# usamos a função solve_ivp para resolver numericamente o problema de valor inicial.
# passamos a função do sistema, o intervalo de integração x de 0 a 10 e o vetor
# com as duas condições iniciais obrigatórias: posição (1) e velocidade (0) agrupadas em [1, 0]
resultExe16 = spin.solve_ivp(sisExe16, (0, 10), [1, 0])

# exibimos os resultados na tela. O atributo '.y[0]' contém a trajetória da posição (y)
# e o atributo '.t' armazena os pontos de tempo calculados pelo integrador
print(f'\nPontos de tempo avaliados: \n{resultExe16.t}')
print(f'Valores calculados para a posição y: \n{resultExe16.y[0]}')
```

20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma
coluna numérica.

```python
# Vintésimo Exercício
import pandas as pd

# criamos um dicionário com os dados do produto
dadosExe20PD = {'Produto': ['p1', 'bp2', 'p3', 'Bp4', 'p5'],
                'Categoria': ['c1', 'c1', 'c2', 'c3', 'c1'],
                'Preço': [10, 20, 30, 35, 40],
               }
# e então transformamos esse dicionário em um Dataframe usando a função
# pd.DataFrame
dfExe20PD = pd.DataFrame(dadosExe20PD)

# usamos a função .groupby para criar uma nova variavel, que será a dataframe
# agrupada por Categoria, com base em Preço e na sua mediana
agrupExe20PD = dfExe20PD.groupby('Categoria')['Preço'].median()
print(agrupExe20PD)
```
